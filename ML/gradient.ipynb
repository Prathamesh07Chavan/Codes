```python
import numpy as np                                      # For numerical calculations
import matplotlib.pyplot as plt                         # For plotting
```

```python
x = 2                                                   # Starting point
learning_rate = 0.1                                     # Set learning rate
iterations = 20                                         # Number of iterations

x_values = [x]                                          # Store x values

for i in range(iterations):                             # Repeat for given iterations
    gradient = 2 * (x + 3)                              # Calculate gradient
    x = x - learning_rate * gradient                    # Update x
    x_values.append(x)                                  # Store updated x

print("Local Minimum x =", x)                           # Display minimum x
print("Minimum y =", (x + 3) ** 2)                     # Display minimum y
```

```python
x_range = np.linspace(-6, 3, 100)                       # Create x values for graph
y_range = (x_range + 3) ** 2                            # Calculate y values

plt.plot(x_range, y_range, label="y = (x + 3)²")       # Plot the function
plt.plot(x_values, [(x + 3) ** 2 for x in x_values],
         "ro-", label="Gradient Descent")               # Plot descent points

plt.xlabel("x")                                         # Label x-axis
plt.ylabel("y")                                         # Label y-axis
plt.title("Gradient Descent for Local Minimum")         # Add graph title
plt.legend()                                             # Show legend
plt.grid(True)                                           # Show grid
plt.show()                                               # Display graph
```
